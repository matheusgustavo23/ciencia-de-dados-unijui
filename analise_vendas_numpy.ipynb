# Análise de Vendas Semanais com NumPy



##Dados

```python
import numpy as np

# matriz 7x5 onde cada linha é um dia e cada coluna é um indicador
dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [ 98, 11700,  82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200,  95, 120.00, 2]
])

dias = np.array(["Segunda", "Terça", "Quarta", "Quinta", "Sexta", "Sábado", "Domingo"])

# indices das colunas
PEDIDOS, FATURAMENTO, CLIENTES, TICKET, RECLAMACOES = 0, 1, 2, 3, 4
indicadores = np.array(["Pedidos", "Faturamento", "Clientes", "Ticket_Medio", "Reclamações"])

# limites usados nos filtros da pt5 e podem ser ajustados
LIMITE_FATURAMENTO = 20000
LIMITE_PEDIDOS = 150
LIMITE_RECLAMACOES = 5

np.set_printoptions(suppress=True)  # n da notação científica
```

## pt1


```python
# .shape devolve (linhas, colunas)
linhas, colunas = dados.shape
print("Qtd de linhas:", linhas)
print("Qtd de colunas:", colunas)
print("Qtd total de valores:", dados.size)
print("Número de dimensões:", dados.ndim)
print("Tipo dos dados:", dados.dtype)
```

pt2


```python
# Linha da sexta-feira (índice 4)
print("Sexta-feira:", dados[4])

# coluna de faturamento (todas as linhas)
print("Faturamento:", dados[:, FATURAMENTO])

# Coluna de reclamações
print("Reclamações:", dados[:, RECLAMACOES])

# Slicing de linhas: quinta (3) até domingo (6)
print("Quinta até domingo:")
print(dados[3:])

# slicing de colunas: pedidos, faturamento e clientes (0 a 2)
print("Pedidos, Faturamento e Clientes:")
print(dados[:, 0:3])
```

pt3

```python
pedidos = dados[:, PEDIDOS]
faturamento = dados[:, FATURAMENTO]
clientes = dados[:, CLIENTES]
reclamacoes = dados[:, RECLAMACOES]

# func. vetorizadas do NumPy
total_pedidos = np.sum(pedidos)
total_faturamento = np.sum(faturamento)
total_clientes = np.sum(clientes)
media_pedidos = np.mean(pedidos)
media_faturamento = np.mean(faturamento)
media_reclamacoes = np.mean(reclamacoes)

print(f"Total de pedidos: {total_pedidos:.0f}")
print(f"Faturamento total: R$ {total_faturamento:,.2f}")
print(f"Total de clientes: {total_clientes:.0f}")
print(f"Média diária de pedidos: {media_pedidos:.2f}")
print(f"Média diária de faturamento: R$ {media_faturamento:,.2f}")
print(f"Média de reclamações por dia: {media_reclamacoes:.2f}")
```

pt4

```python
# argmax/argmin devolvem o indice e usamos ele para obter o nome do dia
i_max_ped = np.argmax(pedidos)
i_min_ped = np.argmin(pedidos)
print(f"Maior nº de pedidos: {dias[i_max_ped]} ({pedidos[i_max_ped]:.0f})")
print(f"Menor nº de pedidos: {dias[i_min_ped]} ({pedidos[i_min_ped]:.0f})")

i_max_fat = np.argmax(faturamento)
i_min_fat = np.argmin(faturamento)
print(f"Maior faturamento: {dias[i_max_fat]} (R$ {faturamento[i_max_fat]:,.2f})")
print(f"Menor faturamento: {dias[i_min_fat]} (R$ {faturamento[i_min_fat]:,.2f})")

# amplitude = maior - menor (np.ptp faz o mesmo cálculo)
amplitude_fat = np.max(faturamento) - np.min(faturamento)
print(f"Amplitude do faturamento: R$ {amplitude_fat:,.2f}")
```

pt5

```python
# masks booleans onde faz a comparação elemento a elemento q gera True/False
mask_fat = faturamento > LIMITE_FATURAMENTO
print(f"Dias com faturamento > R$ {LIMITE_FATURAMENTO:,}:", dias[mask_fat])

mask_ped = pedidos > LIMITE_PEDIDOS
print(f"Dias com mais de {LIMITE_PEDIDOS} pedidos:", dias[mask_ped])

mask_rec = reclamacoes > LIMITE_RECLAMACOES
print(f"Dias com mais de {LIMITE_RECLAMACOES} reclamações:", dias[mask_rec])

# A msm máscara de pedidos aplicada a coluna de faturamento
print("Faturamento dos dias com mais de 150 pedidos:", faturamento[pedidos > 150])

# np.sum em máscara conta os True
abaixo_media = faturamento < media_faturamento
print(f"Dias com faturamento abaixo da média (R$ {media_faturamento:,.2f}): "
      f"{np.sum(abaixo_media)} -> {dias[abaixo_media]}")
```

pt6

```python
# np.where
classificacao = np.where(faturamento >= 20000, "Alto desempenho", "Desempenho normal")

# Empilha dias e classificação lado a lado
print("Classificação por dia:")
print(np.column_stack((dias, classificacao)))

# Conta quantos são "Alto desempenho"
qtd_alto = np.sum(classificacao == "Alto desempenho")
print("\nDias com alto desempenho:", qtd_alto)
```

pt7

```python
media_f = np.mean(faturamento)
mediana_f = np.median(faturamento)
desvio_f = np.std(faturamento)
min_f = np.min(faturamento)
max_f = np.max(faturamento)
cv_f = desvio_f / media_f * 100 # coeficiente de variação (%)

print(f"Média: R$ {media_f:,.2f}")
print(f"Mediana: R$ {mediana_f:,.2f}")
print(f"Desvio padrão: R$ {desvio_f:,.2f}")
print(f"Mínimo: R$ {min_f:,.2f}")
print(f"Máximo: R$ {max_f:,.2f}")
print(f"Coef. de variação: {cv_f:.1f}%")
print(f"O maior faturamento é {max_f / min_f:.2f}x o menor.")

# resposta dos dos valores calculados
nivel = "grande" if cv_f > 20 else "moderada" if cv_f > 10 else "pequena"
print(f"\nResposta: a variação é {nivel}. O desvio padrão (R$ {desvio_f:,.2f}) corresponde a "
      f"{cv_f:.1f}% da média e a amplitude é de R$ {max_f - min_f:,.2f}")
```

sim, tem variação considerável. O desvio padrão é cerca de 26% da média e o melhor dia fatura mais que o dobro do pior. Os valores se dividem em dois blocos onde segunda a quarta e domingo (11–16 mil) e quinta a sábado (
R$ 21–24,5 mil).

```python
# correlação de Pearson entre os indicadores (matriz 3x3)
corr = np.corrcoef(dados[:, [0, 1, 4]], rowvar=False)
print("Correlação entre Pedidos, Faturamento e Reclamações:")
print(np.round(corr, 3))

# Reclamações por 100 pedidos em cada dia
taxa = dados[:, 4] / dados[:, 0] * 100
print("\nReclamações a cada 100 pedidos:")
print(np.column_stack((dias, np.round(taxa, 2))))

# Comparação: dias de alto desempenho x demais
print(f"\nFaturamento médio / alto desempenho: R$ {np.mean(dados[alto, 1]):,.2f}")
print(f"Faturamento médio / demais dias:      R$ {np.mean(dados[~alto, 1]):,.2f}")
print(f"Participação de quinta a sábado no faturamento: "
      f"{np.sum(dados[alto, 1]) / np.sum(dados[:, 1]) * 100:.1f}%")
```
