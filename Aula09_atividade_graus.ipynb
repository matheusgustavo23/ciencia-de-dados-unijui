```python
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from time import sleep
```

```python
uf = input("Digite o nome da UF:")
mun = input("Digite o nome da cidade: ")
```

```python
url = f"https://hgbrasil.com/weather/places/brazil/{uf}/{mun}"
navegador = gs.Chrome()
```

```python
import re
def remove_non_numeric_from_str(value: str):
  return re.sub("[^0-9]", "", value)
```

```python
navegador.get(url)
sleep(2)
print("Pag: ", navegador.title)
```

```python
max = remove_non_numeric_from_str(
    navegador.find_element(
      By.XPATH,
      '//*["@id=main"]/article/div/div/section[1]/div[4]/div[2]/span[1]'
).text)
```

```python
min = remove_non_numeric_from_str(
    navegador.find_element(
      By.XPATH,
      '//*[@id="main"]/article/div/div/section[1]/div[4]/div[2]/span[2]'
).text)
```

```python
print(f"Temperatura em {mun} ({uf}): ")
print(f"Maxima: {max} °C")
print(f"Minima: {min} °C")
```
