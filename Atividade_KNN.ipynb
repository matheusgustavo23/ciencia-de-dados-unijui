```python
import pandas as pd
import numpy as np

colunas = ['Acidez Fixa','Acidez Volátil','Acidez Cítrica','Açucar Residual','Cloreto','Livre de Dióxido de Enxofre','Dióxido de Enxofre Total','Densidade','pH','Sulfato','Álcool','Qualidade']
df_vinhos = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv', names = colunas, sep = ';', skiprows=1)
df_vinhos.head()
```

```python
condicoes = [df_vinhos['Qualidade'] >= 7.0, df_vinhos['Qualidade'] >= 5.0, df_vinhos['Qualidade'] < 5.0]
notas = ['Alta', 'Média', 'Baixa']

df_vinhos['Classificação'] = np.select(condicoes, notas, default = 'Desconhecido')
df_vinhos.head(n = 10)
```

```python
X = df_vinhos[df_vinhos.columns.difference(['Classificação', 'Qualidade'])].values
print(f'X: {X}')

y = df_vinhos['Classificação'].values
print(f'Y: {y}')
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score

vinho_classificador = KNeighborsClassifier(n_neighbors = 5)
vinho_classificador.fit(X, y)
```

```python
score_dt = cross_val_score(vinho_classificador, X, y, scoring = 'accuracy', cv =5)
print(score_dt.mean())
```

```python
for i in range(1, 11):
    vinho_classificador = KNeighborsClassifier(n_neighbors = i)
    vinho_classificador.fit(X, y)
    score_dt = cross_val_score(vinho_classificador, X, y, scoring='accuracy', cv=5)
    print(f'Vizinhos: {i} Acurácia: {score_dt.mean()}')
```
