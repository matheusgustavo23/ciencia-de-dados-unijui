```python
import requests
import json
```

```python
ceps = [
    # sr
    "98780-558", "98780-196", "98780-001", "98780-060", "98780-100", "98780-103",
    "98783-110", "98785-253", "98796-050", "98794-298",
    "98790-068", "98792-710", "98780-732", "98792-590", "98780-325",
    # tm, hz, giruá
    "98910-000", "98922-970", "98870-000",
    # santo angelo
    "98805-816", "98807-039", "98805-807", "98804-867",
    # cruz alta
    "98050-370", "98015-060", "98010-620", "98015-430",
    # erro
    "99999-999"
]

registros = [] # lista onde vai guardar os resultados

for cep in ceps:
  url = f"https://viacep.com.br/ws/{cep}/json/"
  resposta = requests.get(url, timeout = 10)
  dados = resposta.json()
  if "erro" not in dados:
    registros.append(dados)
  else:
    print(f"CEP {cep} não encontrado")

print(len(registros))
```

```python
print(registros[0])
```

```python
colunas = ["cep", "logradouro", "bairro", "localidade", "ibge", "uf", "ddd"]

dados_limpos = []

for registro in registros:
  linha = {}
  for coluna in colunas:
    linha[coluna] = registro[coluna]
  dados_limpos.append(linha)

print(dados_limpos[4])
```

```python
with open("ceps_regiao.json", "w", encoding="utf-8") as arquivo:
    json.dump(dados_limpos, arquivo, ensure_ascii=False, indent=2)
```
